```python
import os
import numpy as np
import torch as t
from sae_lens import (
    SAE,
    HookedSAETransformer,
)
import gc
import pickle
from datasets import load_dataset
```

```python
from huggingface_hub import notebook_login
device = "cuda"
t.set_grad_enabled(False)
notebook_login()
```

### Load Model

```python
!pip install hf_transfer
model_ver = "2b"
if model_ver == "2b":
    total_layers = 26
else:
    total_layers = 42
gemma: HookedSAETransformer = HookedSAETransformer.from_pretrained(f"gemma-2-{model_ver}", device=device)
tokenizer = gemma.tokenizer
```

### Load dataset

```python
# Load the GoEmotions dataset
dataset = load_dataset("mrm8488/goemotions")

# Get the train split
data = dataset['train']
```

```python
# Sample and prepare the data
print("Starting data preparation...")
np.random.seed(42)

# Filter texts by emotion labels
all_neutral = [example['text'] for example in data if example['neutral'] == 1]
all_anger = [example['text'] for example in data if example['anger'] == 1]

print(f"Total neutral sentences: {len(all_neutral)}")
print(f"Total anger sentences: {len(all_anger)}")

# Sample 1000 from each (or max available)
num_samples = min(1000, len(all_neutral), len(all_anger))
print(f"\nSampling {num_samples} examples from each")

neutral_indices = np.random.choice(len(all_neutral), size=num_samples, replace=False)
anger_indices = np.random.choice(len(all_anger), size=num_samples, replace=False)

neutral_prompts = [all_neutral[i] for i in neutral_indices]
anger_prompts = [all_anger[i] for i in anger_indices]

# Check first examples
print(f"\nNeutral example:")
print(neutral_prompts[0])
print(f"\nAnger example:")
print(anger_prompts[0])
print("\nData preparation complete!")

# Calculate and save token lengths
neutral_token_length = [tokenizer(p, return_tensors="pt").input_ids.shape[1] for p in neutral_prompts]
anger_token_length = [tokenizer(p, return_tensors="pt").input_ids.shape[1] for p in anger_prompts]
print(f"\nAverage neutral token length: {np.mean(neutral_token_length):.1f}")
print(f"Average anger token length: {np.mean(anger_token_length):.1f}")

# Save token lengths
os.makedirs("./token_lengths", exist_ok=True)
with open("./token_lengths/emotion_neutral.pkl", "wb") as f:
    pickle.dump(neutral_token_length, f)

with open("./token_lengths/emotion_anger.pkl", "wb") as f:
    pickle.dump(anger_token_length, f)
print("\n✅ Token lengths saved to ./token_lengths/")
```

### Retrieve SAE Activations

```python
sae_release = f"gemma-scope-{model_ver}-pt-res-canonical"
sae_id = "layer_0/width_16k/canonical"

blobs_path = f"/root/.cache/huggingface/hub/models--google--gemma-scope-{model_ver}-pt-res/blobs"

# Create output directories
os.makedirs(f"./sae_acts_{model_ver}/goemotions", exist_ok=True)

# SMALLER BATCH SIZE: Process 10 examples at a time
batch_size = 10
num_batches = 100  # 10 * 100 = 1000 total

for i in range(total_layers):
    sae = SAE.from_pretrained(
        release=sae_release,
        sae_id=f"layer_{i}/width_16k/canonical",
        device=str(device),
    )[0]  # Get just the SAE object

    for j in range(num_batches):
        start_idx = j * batch_size
        end_idx = (j + 1) * batch_size

        # Process neutral sentences
        _, cache = gemma.run_with_cache_with_saes(neutral_prompts[start_idx:end_idx], saes=[sae])
        sae_acts_post = cache[f"{sae.cfg.metadata.get('hook_name')}.hook_sae_acts_post"].cpu()
        t.save(sae_acts_post, f"./sae_acts_{model_ver}/goemotions/neutral_layer{i}_idx{end_idx}.pt")

        cache.to("cpu")
        del cache, sae_acts_post
        t.cuda.empty_cache()
        gc.collect()

        # Clean up blobs
        if os.path.exists(blobs_path):
            for item in os.listdir(blobs_path):
                item_path = os.path.join(blobs_path, item)
                if os.path.isfile(item_path):
                    os.remove(item_path)

        # Process anger sentences
        _, cache = gemma.run_with_cache_with_saes(anger_prompts[start_idx:end_idx], saes=[sae])
        sae_acts_post = cache[f"{sae.cfg.metadata.get('hook_name')}.hook_sae_acts_post"].cpu()
        t.save(sae_acts_post, f"./sae_acts_{model_ver}/goemotions/anger_layer{i}_idx{end_idx}.pt")

        cache.to("cpu")
        del cache, sae_acts_post
        t.cuda.empty_cache()
        gc.collect()

        # Clean up blobs
        if os.path.exists(blobs_path):
            for item in os.listdir(blobs_path):
                item_path = os.path.join(blobs_path, item)
                if os.path.isfile(item_path):
                    os.remove(item_path)

        # Print progress every 10 batches
        if (j + 1) % 10 == 0:
            print(f"  Layer {i}: {j+1}/{num_batches} batches completed")

    sae.cpu()
    del sae
    t.cuda.empty_cache()
    gc.collect()
    print(f"✅ Completed layer {i}")

print(f"\n🎉 All files saved to: ./sae_acts_{model_ver}/goemotions/")
```
